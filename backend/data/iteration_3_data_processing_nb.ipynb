## Summary

The purpose of this notebook is to process the ALA Yearly Occurrences Data for Wildlife Seminary Feature 7 (Animal Timeline Comparisons).

Data already includes conservation status and latitude/longitude joined onto yearly occurrences.

```python
import pandas as pd
pd.set_option('display.max_rows', 10000)
```

### Reading in Processed Data

```python
df = pd.read_csv("wildlife_occurrences_with_location_end_status_v2_w_lat_long.csv")
```

```python
display(df.head())
df.shape
```

### Starting from 1970

```python
df = df.query("year >= 1970 ").reset_index(drop = True)

# remove extra column
df = df.drop("CommonName", axis = 1)
```

```python
df.head()
```

```python
df.head()
```

### Aggregating Datasets for Animal Counts by Year, State, EPBCThreat Status, Kingdom

```python
agg_cols = ["year","stateTerritory","ScientificName", "CommonName_clean", "EPBCThreatStatus"]

df_agg = (df.groupby(agg_cols).
 agg({"occurrenceCount": "sum"})).reset_index().rename(columns = {"CommonName_clean": "CommonName"})
```

```python
df_agg.head()
```

```python
# converting to csv
df_agg.to_csv("epic_6_agg_wildlife_occurrences_1970_2022.csv")
```

### Creating Dataset with Names and Wikipedia Link

```python
top_animals_df = pd.read_csv("animal_details_top_animals.csv")
top_animals_df = top_animals_df[["CommonName", "ScientificName", "occurrenceCount"]]
```

```python
import urllib.parse

# Create a Wikipedia link column
top_animals_df["wiki_link"] = top_animals_df["ScientificName"].apply(
    lambda name: f"https://en.wikipedia.org/wiki/{urllib.parse.quote(name.replace(' ', '_'))}"
)

# Display result
top_animals_df
```
